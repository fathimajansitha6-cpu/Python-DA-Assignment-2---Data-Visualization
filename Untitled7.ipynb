```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#Getting Tools & Data: We bring in our charting tools (Pandas, Matplotlib, Seaborn) and download the taxi dataset.

# 1) Loading the Taxis Dataset
# Load the 'taxis' dataset
df = sns.load_dataset("taxis")

# 2) Handling Missing Values
print("Missing values before cleaning:\n", df.isnull().sum())

# Impute missing values using mode for categorical columns
categorical_cols = ['payment', 'pickup_zone', 'dropoff_zone', 'pickup_borough', 'dropoff_borough']
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# Impute missing numerical values using the median (if any were missing)[cite: 1]
numerical_cols = df.select_dtypes(include=['float64', 'int64']).columns
for col in numerical_cols:
    df[col] = df[col].fillna(df[col].median())

# For columns that are critical and cannot be reasonably imputed (like timestamps), remove those rows[cite: 1]
df = df.dropna()

print("\nMissing values after cleaning:\n", df.isnull().sum())

#Fixing Blanks: Charts crash if there is missing data. We tell the computer to guess the missing words (using the most common word) and missing numbers (using the middle number), and we delete rows that are too broken to fix.
```

```python
# Convert pickup to datetime format before plotting[cite: 1]
df['pickup'] = pd.to_datetime(df['pickup'])

# Sort data by time so the line chart connects points in chronological order
df_sorted = df.sort_values('pickup')

# Line Chart: Visualize fare over time (pickup timestamp on x-axis, fare on y-axis)[cite: 1]
plt.figure(figsize=(12, 5))
plt.plot(df_sorted['pickup'], df_sorted['fare'], alpha=0.7)
plt.title('Fare Over Time')
plt.xlabel('Pickup Time')
plt.ylabel('Fare')
plt.show()
#Line Chart: Connects the dots to show how taxi fares go up and down over time
```

```python
# Bar Chart: Total fare for each pickup_borough[cite: 1]
# Group data by pickup_borough and sum the fare[cite: 1]
borough_fare = df.groupby('pickup_borough')['fare'].sum()
borough_fare.plot(kind='bar', figsize=(8, 5), color='skyblue', edgecolor='black')
plt.title('Total Fare by Pickup Borough')
plt.xlabel('Pickup Borough')
plt.ylabel('Total Fare')
plt.show()
#Bar Chart: Adds up all the fares for each neighborhood and compares them side-by-side as bars.
```

```python
# Pie Chart: Distribution of trips based on payment method[cite: 1]
payment_counts = df['payment'].value_counts()
payment_counts.plot(kind='pie', autopct='%1.1f%%', figsize=(6, 6), startangle=90)
plt.title('Trips by Payment Method')
plt.ylabel('') # Hides the default y-axis label
plt.show()
#Pie Chart: Slices a circle to show the percentage of people paying with cash versus credit cards.
```

```python
# Histogram: Visualize the distribution of distance with customized bins[cite: 1]
plt.figure(figsize=(8, 5))
plt.hist(df['distance'], bins=30, edgecolor='black', color='salmon')
plt.title('Distribution of Trip Distance')
plt.xlabel('Distance')
plt.ylabel('Frequency')
plt.show()
#Histogram: Groups trips by distance (e.g., 0-5 miles, 5-10 miles) so you can see if most taxi rides are short or long.
```

```python
# Box Plot: Tip amounts for each pickup_borough[cite: 1]
# pickup_borough as categorical axis, tip as numeric axis[cite: 1]
df.boxplot(column='tip', by='pickup_borough', figsize=(8, 5), grid=False)
plt.title('Distribution of Tip Amounts by Pickup Borough')
plt.suptitle('') # Removes pandas' default title
plt.xlabel('Pickup Borough')
plt.ylabel('Tip Amount')
plt.show()
#Box Plot: Shows the spread of tips. The box shows what an "average" tip looks like, and the dots show the crazy high or low tips.
```

```python
# Count Plot: Number of trips in each pickup_borough[cite: 1]
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='pickup_borough', palette='viridis')
plt.title('Number of Trips in Each Pickup Borough')
plt.xlabel('Boroughs')
plt.ylabel('Count of Trips')
plt.show()
#Count Plot: A faster, automatic bar chart that just counts how many trips started in each neighborhood.
```

```python
# Scatter Plot: Relationship between distance (x-axis) and fare (y-axis)[cite: 1]
# Color points based on pickup_borough[cite: 1]
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='distance', y='fare', hue='pickup_borough', alpha=0.6)
plt.title('Distance vs. Fare by Pickup Borough')
plt.show()
#Scatter Plot: Puts a dot on the screen for every single trip to see if longer distances always equal higher fares
```

```python
# Heatmap: Correlation between distance, fare, tip, tolls, and total[cite: 1]
plt.figure(figsize=(8, 6))
num_vars = ['distance', 'fare', 'tip', 'tolls', 'total']
# Use a correlation matrix to highlight relationships[cite: 1]
corr_matrix = df[num_vars].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
#Heatmap: Uses a color scale (like a weather map) to show how strongly things are linked. Darker colors mean things like "distance" and "total price" rise together.
```

```python
# Pair Plot: Pairwise relationships between distance, fare, tip, and total[cite: 1]
# Color data points according to pickup_zone[cite: 1]
# Note: pickup_zone has many categories, so the legend might be large.
sns.pairplot(df[['distance', 'fare', 'tip', 'total', 'pickup_zone']], hue='pickup_zone')
plt.show()
#Pair Plot: An automatic grid that creates a scatter plot for every possible combination of your numbers all at once.
```

```python
# Violin Plot: Distribution of fare for each payment method[cite: 1]
# Payment method as categorical axis, fare as numeric axis[cite: 1]
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='payment', y='fare', palette='muted')
plt.title('Distribution of Fare by Payment Method')
plt.show()
#Violin Plot: Looks like a guitar or violin. The fatter the shape is at a certain dollar amount, the more people paid that exact amount.
```
